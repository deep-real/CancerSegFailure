# MedFoE concept localization

Use the shared segmentation checkpoint and layer-11, step-6200 SAE. Set one prepared MRI slice below. No reference mask is used.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "predict.py").is_file():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
CONFIG = ROOT / "configs/picai.json"
IMAGE = ROOT / "data/picai/slices/PATIENT_STUDY/0.hdf5"  # Set your prepared MRI slice.
DEVICE = "cuda:0"  # Or "cpu".
```

```python
from medfoe.concepts import localize, plot_concepts, plot_revision

result = localize(CONFIG, IMAGE, concept_ids=[469, 968, 204, 1097],
                  layer=11, step=6200, device=DEVICE, revise_with=469)
plot_concepts(result);
```

```python
plot_revision(result);
```

The heatmaps show trained SAE responses. The single-concept reconstruction can add or remove pixels; it is a qualitative example, separate from the full prediction command. Change the concept IDs to inspect other features.
